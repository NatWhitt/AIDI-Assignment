```python
import pandas as pd 
```

```python
# 1. Load your CSV file
df = pd.read_csv('Assignment data premessage.csv')
df.head()
```

```python
from numpy import random
def format_message(row):
  if random.randint(0, 2) == 0:
    return f"{row['Type']} detention set for {row['Student']} on {row['Date']} at {row['Time']}. Reason: {row['Reason']}",
  elif random.randint(0, 2) == 1:
    return f"{row['Student']} has been set a detention for {row['Type']} on {row['Date']} at {row['Time']}. For the reason of {row['Reason']}",
  elif random.randint(0, 2) == 2:
    return f"Add {row['Student']} into a detention {row['Type']} on {row['Date']} at {row['Time']}. For the reason of {row['Reason']}",
```

```python
df['Message'] = df.apply(
      lambda row: f"{row['Type']} detention set for {row['Student']} on {row['Date']} at {row['Time']}. For {row['Reason']}",
      axis=1
)
```
